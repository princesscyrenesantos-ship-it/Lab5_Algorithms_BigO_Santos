```python
# [CELL 3] LINEAR GROWTH
def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

```python
# [CELL 4] BIG O CLASSIFICATION
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10, 100, 1000]:
    print(
        f"n={n:4d} | O(1)=1 | "
        f"O(log n)={log_ops(n):2d} | "
        f"O(n)={n:4d} | O(n^2)={n*n:7d}"
    )
```

```python
# [CELL 5] SPACE COMPLEXITY
def constant_space(n):
    total = 0
    for value in range(n):
        total += value
    return total

def linear_space(n):
    return list(range(n))

for n in [10, 100, 1000, 5000]:
    data = linear_space(n)
    print(f"n={n:5d} | O(1): constant | O(n): {len(data)}")
```

```python
# [CELL 6] CASE ANALYSIS
def search_case(data, target):
    for i, value in enumerate(data, 1):
        if value == target:
            return i
    return len(data)

data = [11, 22, 33, 44, 55]
for case, target in {"Best": 11, "Average": 33, "Worst": 99}.items():
    print(f"{case} Case: {search_case(data, target)} comparisons")
```

```python
# [CELL 7] LINEAR SEARCH
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)
print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 8] BINARY SEARCH
def binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons

index, comparisons = binary_search(data, target)
print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 9] SEARCH COMPARISON
data = list(range(1, 101))
target = ((ID_SUM + SEED_DIGIT) % 100) + 1

li, lc = linear_search(data, target)
bi, bc = binary_search(data, target)

print("Target:", target)
print(f"Linear -> index={li}, comparisons={lc}")
print(f"Binary -> index={bi}, comparisons={bc}")
print("Reduction:", lc - bc)
```

```python
# [CELL 10] BUBBLE SORT
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

data = [
    (ID_SUM * 3) % 100, (SEED_DIGIT * 11) % 100,
    (NAME_LENGTH * 7) % 100, 42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps = bubble_sort(data)
print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# [CELL 11] SELECTION SORT
def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

sorted_data, comparisons, swaps = selection_sort(data)
print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# [CELL 12] INSERTION SORT
def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

sorted_data, comparisons, shifts = insertion_sort(data)
print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

```python
# [CELL 13] EXECUTION-TIME MEASUREMENT
import time

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

data_size = 100 + ID_SUM * 10
performance_data = [
    (ID_SUM * 3 + i * 7) % 100
    for i in range(data_size)
]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print(f"Dataset Size: {data_size}")
print(f"Bubble Sort: {bubble_time:.8f} s")
print(f"Selection Sort: {selection_time:.8f} s")
print(f"Insertion Sort: {insertion_time:.8f} s")
```

```python
# [CELL 14] SORTING PERFORMANCE COMPARISON
results = [
    ("Bubble Sort", bubble_result, bubble_time),
    ("Selection Sort", selection_result, selection_time),
    ("Insertion Sort", insertion_result, insertion_time)
]

print(f"{'Algorithm':<18}{'Comparisons':>12}{'Movements':>12}{'Time (s)':>12}")
for name, result, elapsed in results:
    print(
        f"{name:<18}{result[1]:>12}"
        f"{result[2]:>12}{elapsed:>12.8f}"
    )
```

```python
# EXERCISE 1: ROTATED RECORD SEARCH
orig_records = [10, 20, 30, 40, 50, 60, 70, 80]
seed_digit = 6  # SEED_DIGIT
rotation_point = seed_digit % len(orig_records)
rotated_records = orig_records[rotation_point:] + orig_records[:rotation_point]
target_incident = 50

def linear_search_rotated(arr, target):
    comps = 0
    for i, val in enumerate(arr):
        comps += 1
        if val == target:
            return i, comps
    return -1, comps

def binary_search_rotated(arr, target):
    low, high, comps = 0, len(arr) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comps += 1
        if arr[mid] == target:
            return mid, comps
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comps

l_idx, l_comps = linear_search_rotated(rotated_records, target_incident)
b_idx, b_comps = binary_search_rotated(rotated_records, target_incident)

print("=== EXERCISE 1: EMERGENCY DISPATCH SEARCH ===")
print("Original Dataset:", orig_records)
print("Rotated Dataset:", rotated_records)
print("Target Incident:", target_incident)
print(f"Linear Search -> Index: {l_idx}, Comparisons: {l_comps}")
print(f"Modified Binary Search -> Index: {b_idx}, Comparisons: {b_comps}")
print("Complexity -> Time: O(log n) | Space: O(1)")
```

```python
# EXERCISE 2: CONSOLIDATED ENROLLMENT RECORDS
dept_a = [6, 12, 18, 24, 30]
dept_b = [5, 11, 17, 23, 29]

def consolidate_records(arr1, arr2):
    i = j = comps = 0
    merged = []
    while i < len(arr1) and j < len(arr2):
        comps += 1
        if arr1[i] <= arr2[j]:
            merged.append(arr1[i])
            i += 1
        else:
            merged.append(arr2[j])
            j += 1
    while i < len(arr1):
        merged.append(arr1[i])
        i += 1
    while j < len(arr2):
        merged.append(arr2[j])
        j += 1
    return merged, comps

consolidated, merge_comps = consolidate_records(dept_a, dept_b)

print("\n=== EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS ===")
print("Department A:", dept_a)
print("Department B:", dept_b)
print("Consolidated Dataset:", consolidated)
print(f"Records Before: {len(dept_a) + len(dept_b)} | After: {len(consolidated)}")
print(f"Comparison Count: {merge_comps}")
print(f"Ordering Verification: {consolidated == sorted(consolidated)}")
print("Complexity -> Time: O(n + m) | Space: O(n + m)")
```
